This Jupyter notebook covers the core subsetting, filtering, and data manipulation techniques in Pandas using the **Titanic dataset**.

## 1. Downloading & Loading the Dataset

The dataset used in this tutorial is `titanic.csv` from the official pandas documentation repository.

- Direct URL: `https://raw.githubusercontent.com/pandas-dev/pandas/main/doc/data/titanic.csv`

You can load it directly from the web or from a locally downloaded file using `pd.read_csv()`.

```python
import pandas as pd

# Load the dataset from local file (or replace with the direct URL)
titanic = pd.read_csv("titanic.csv")

# Display the first 5 rows
titanic.head()
```

### Inspecting the Dataset
Use `.info()` to inspect column names, data types, and check for missing/null values.

```python
titanic.info()
```

## 2. Selecting Specific Columns

- **Single column**: `titanic["Age"]` returns a `pandas.Series`.
- **Multiple columns**: `titanic[["Age", "Sex"]]` returns a `pandas.DataFrame`.

```python
# Selecting a single column
ages = titanic["Age"]
print("Type:", type(ages))
print("Shape:", ages.shape)
ages.head()
```

```python
# Selecting multiple columns
age_sex = titanic[["Age", "Sex"]]
print("Type:", type(age_sex))
print("Shape:", age_sex.shape)
age_sex.head()
```

## 3. Filtering Specific Rows (Boolean Indexing)

Filter rows by passing conditional statements inside square brackets `df[condition]`.

### 3.1 Numeric Filtering
Filter passengers older than 35 years.

```python
above_35 = titanic[titanic["Age"] > 35]
print(f"Number of passengers older than 35: {len(above_35)}")
above_35.head()
```

### 3.2 Filtering with `.isin()`
Select passengers traveling in cabin classes 2 and 3.

```python
class_2_3 = titanic[titanic["Pclass"].isin([2, 3])]
print(f"Passengers in Class 2 or 3: {len(class_2_3)}")
class_2_3.head()
```

### 3.3 Handling Missing Data with `.notna()`
Filter out rows where the `Age` value is missing (`NaN`).

```python
age_known = titanic[titanic["Age"].notna()]
print(f"Passengers with known age: {len(age_known)} / {len(titanic)}")
age_known.head()
```

## 4. Selecting Specific Rows and Columns (`.loc` and `.iloc`)

- **`.loc`** (label-based): `df.loc[row_condition, [column_names]]`
- **`.iloc`** (integer position-based): `df.iloc[row_range, column_range]`

### 4.1 Using `.loc`
Get names of passengers older than 35.

```python
adult_names = titanic.loc[titanic["Age"] > 35, "Name"]
adult_names.head()
```

### 4.2 Using `.iloc`
Select rows 10 to 25 and columns 3 to 5 (0-indexed).

```python
subset = titanic.iloc[9:25, 2:5]
subset
```

## 5. Summary of Key Pandas Subsetting Methods

| Task | Syntax | Output |
| :--- | :--- | :--- |
| Single Column | `df["col"]` | Series |
| Multiple Columns | `df[["col1", "col2"]]` | DataFrame |
| Filter Rows | `df[df["col"] > val]` | DataFrame |
| Set Membership | `df[df["col"].isin([a, b])]` | DataFrame |
| Filter Nulls | `df[df["col"].notna()]` | DataFrame |
| Condition + Columns | `df.loc[condition, ["cols"]]` | Series / DataFrame |
| Slice by Index | `df.iloc[row_start:row_end, col_start:col_end]` | DataFrame |
